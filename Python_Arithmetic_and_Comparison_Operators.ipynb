# Python Arithmetic and Comparison Operators

Simple practice examples using basic Python data types. No `try-except` blocks are used.

# ADDITION OPERATION

Using the `+` arithmetic operator.

```python
# int
a = 10
b = 5
print(a + b)

# float
a = 10.5
b = 2.5
print(a + b)

# bool
a = True
b = False
print(a + b)

# str
a = "Python "
b = "Programming"
print(a + b)

# list
a = [1, 2]
b = [3, 4]
print(a + b)

# tuple
a = (1, 2)
b = (3, 4)
print(a + b)

```

# SUBTRACTION OPERATION

Using the `-` arithmetic operator.

```python
# int
a = 10
b = 5
print(a - b)

# float
a = 10.5
b = 2.5
print(a - b)

# bool
a = True
b = False
print(a - b)

```

# MULTIPLICATION OPERATION

Using the `*` arithmetic operator.

```python
# int
a = 10
b = 5
print(a * b)

# float
a = 10.5
b = 2.5
print(a * b)

# bool
a = True
b = False
print(a * b)

# str
a = "Python"
b = 3
print(a * b)

# list
a = [1, 2]
b = 3
print(a * b)

# tuple
a = (1, 2)
b = 3
print(a * b)

```

# DIVISION OPERATION

Using the `/` arithmetic operator.

```python
# int
a = 10
b = 5
print(a / b)

# float
a = 10.5
b = 2.5
print(a / b)

# bool
a = True
b = 1
print(a / b)

```

# MODULUS OPERATION

Using the `%` arithmetic operator.

```python
# int
a = 10
b = 3
print(a % b)

# float
a = 10.5
b = 3.0
print(a % b)

# bool
a = True
b = 1
print(a % b)

```

# FLOOR DIVISION OPERATION

Using the `//` arithmetic operator.

```python
# int
a = 10
b = 3
print(a // b)

# float
a = 10.5
b = 2.0
print(a // b)

# bool
a = True
b = 1
print(a // b)

```

# GREATER THAN OPERATION

Using the `>` comparison operator.

```python
# int
a = 10
b = 5
print(a > b)

# float
a = 10.5
b = 5.5
print(a > b)

# bool
a = True
b = False
print(a > b)

# str
a = "Python"
b = "Java"
print(a > b)

```

Using the `<` comparison operator.

```python
# int
a = 10
b = 5
print(a < b)

# float
a = 10.5
b = 5.5
print(a < b)

# bool
a = True
b = False
print(a < b)

# str
a = "Python"
b = "Java"
print(a < b)

```

Using the `<=` comparison operator.

```python
# int
a = 10
b = 5
print(a <= b)

# float
a = 10.5
b = 5.5
print(a <= b)

# bool
a = True
b = False
print(a <= b)

# str
a = "Python"
b = "Java"
print(a <= b)

```

# EQUAL TO OPERATION

Using the `==` comparison operator.

```python
# int
a = 10
b = 5
print(a == b)

# float
a = 10.5
b = 5.5
print(a == b)

# bool
a = True
b = False
print(a == b)

# str
a = "Python"
b = "Java"
print(a == b)

```

# NOT EQUAL TO OPERATION

Using the `!=` comparison operator.

```python
# int
a = 10
b = 5
print(a != b)

# float
a = 10.5
b = 5.5
print(a != b)

# bool
a = True
b = False
print(a != b)

# str
a = "Python"
b = "Java"
print(a != b)

```

# Important Note

Different operators are supported by different data types. For example, `+` can add numbers or concatenate strings/lists/tuples, while `-`, `/`, `%`, and `//` are mainly demonstrated here with numeric values.
