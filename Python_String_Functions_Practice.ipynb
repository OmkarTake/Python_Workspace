## capitalize()

**Definition:** Returns a copy of the string with the first character capitalized and the remaining characters converted to lowercase.

```python
s = "welcome TO python"
print(s.capitalize())
```

## casefold()

**Definition:** Returns a lowercase version of the string suitable for case-insensitive comparisons.

```python
s = "PYTHON"
print(s.casefold())
```

## center()

**Definition:** Returns the string centered within a specified width, optionally padded with a chosen character.

```python
s = "Python"
print(s.center(20, "-"))
```

## count()

**Definition:** Returns the number of non-overlapping occurrences of a substring in the string.

```python
s = "banana"
print(s.count("a"))
```

## encode()

**Definition:** Encodes the string into bytes using the specified encoding.

```python
s = "Python"
print(s.encode())
```

## endswith()

**Definition:** Checks whether the string ends with a specified suffix and returns True or False.

```python
s = "Welcome to Python"
print(s.endswith("Python"))
```

## expandtabs()

**Definition:** Replaces tab characters with spaces according to the specified tab size.

```python
s = "Python\tProgramming"
print(s.expandtabs(4))
```

## find()

**Definition:** Returns the lowest index where a substring is found, or -1 if it is not found.

```python
s = "Welcome to Python"
print(s.find("Python"))
```

## format()

**Definition:** Formats a string by replacing placeholders with supplied values.

```python
name = "Omkar"
age = 21
print("My name is {} and I am {} years old.".format(name, age))
```

## index()

**Definition:** Returns the lowest index where a substring is found. It raises ValueError if the substring is not found.

```python
s = "Welcome to Python"
print(s.index("Python"))
```

## isalnum()

**Definition:** Returns True if all characters are alphanumeric and there is at least one character.

```python
s = "Python123"
print(s.isalnum())
```

## isalpha()

**Definition:** Returns True if all characters are alphabetic and there is at least one character.

```python
s = "Python"
print(s.isalpha())
```

## isascii()

**Definition:** Returns True if all characters in the string are ASCII characters.

```python
s = "Python123"
print(s.isascii())
```

## isdecimal()

**Definition:** Returns True if all characters are decimal characters and there is at least one character.

```python
s = "12345"
print(s.isdecimal())
```

## isdigit()

**Definition:** Returns True if all characters are digits and there is at least one character.

```python
s = "12345"
print(s.isdigit())
```

## isidentifier()

**Definition:** Returns True if the string is a valid Python identifier.

```python
s = "student_name"
print(s.isidentifier())
```

## islower()

**Definition:** Returns True if all cased characters are lowercase and there is at least one cased character.

```python
s = "hello python"
print(s.islower())
```

## isnumeric()

**Definition:** Returns True if all characters are numeric characters and there is at least one character.

```python
s = "12345"
print(s.isnumeric())
```

## isprintable()

**Definition:** Returns True if all characters are printable or the string is empty.

```python
s = "Hello Python!"
print(s.isprintable())
```

## isspace()

**Definition:** Returns True if all characters are whitespace characters and there is at least one character.

```python
s = "   "
print(s.isspace())
```

## istitle()

**Definition:** Returns True if the string follows title-case rules.

```python
s = "Welcome To Python"
print(s.istitle())
```

## isupper()

**Definition:** Returns True if all cased characters are uppercase and there is at least one cased character.

```python
s = "PYTHON"
print(s.isupper())
```

## join()

**Definition:** Joins the elements of an iterable using the string as the separator.

```python
words = ["Python", "is", "easy"]
print(" ".join(words))
```

## ljust()

**Definition:** Returns a left-justified string padded to a specified width.

```python
s = "Python"
print(s.ljust(10, "-"))
```

## lower()

**Definition:** Returns a copy of the string converted to lowercase.

```python
s = "PYTHON PROGRAMMING"
print(s.lower())
```

## lstrip()

**Definition:** Returns a copy of the string with leading whitespace or specified characters removed.

```python
s = "   Python"
print(s.lstrip())
```

## partition()

**Definition:** Splits the string at the first occurrence of a separator and returns a three-part tuple.

```python
s = "Welcome to Python"
print(s.partition("to"))
```

## replace()

**Definition:** Returns a copy of the string with specified occurrences replaced by another substring.

```python
s = "I like Java"
print(s.replace("Java", "Python"))
```

## rfind()

**Definition:** Returns the highest index where a substring is found, or -1 if it is not found.

```python
s = "banana"
print(s.rfind("a"))
```

## rindex()

**Definition:** Returns the highest index where a substring is found. It raises ValueError if not found.

```python
s = "banana"
print(s.rindex("a"))
```

## rjust()

**Definition:** Returns a right-justified string padded to a specified width.

```python
s = "Python"
print(s.rjust(10, "-"))
```

## rpartition()

**Definition:** Splits the string at the last occurrence of a separator and returns a three-part tuple.

```python
s = "one-two-three"
print(s.rpartition("-"))
```

## rsplit()

**Definition:** Splits the string from the right and returns a list of substrings.

```python
s = "one-two-three"
print(s.rsplit("-", 1))
```

## rstrip()

**Definition:** Returns a copy of the string with trailing whitespace or specified characters removed.

```python
s = "Python   "
print(s.rstrip())
```

## split()

**Definition:** Splits the string into a list of substrings using a separator or whitespace by default.

```python
s = "Python is easy"
print(s.split())
```

## splitlines()

**Definition:** Splits the string at line boundaries and returns a list of lines.

```python
s = "Python\nJava\nC++"
print(s.splitlines())
```

## startswith()

**Definition:** Checks whether the string starts with a specified prefix and returns True or False.

```python
s = "Welcome to Python"
print(s.startswith("Welcome"))
```

## strip()

**Definition:** Returns a copy of the string with leading and trailing whitespace or specified characters removed.

```python
s = "   Python   "
print(s.strip())
```

## swapcase()

**Definition:** Returns a copy of the string with uppercase characters converted to lowercase and vice versa.

```python
s = "Hello PYTHON"
print(s.swapcase())
```

## title()

**Definition:** Returns a copy of the string with the first character of each word converted to uppercase.

```python
s = "welcome to python"
print(s.title())
```

## translate()

**Definition:** Returns a translated copy of the string using a translation table.

```python
table = str.maketrans("ae", "12")
s = "apple"
print(s.translate(table))
```

## upper()

**Definition:** Returns a copy of the string converted to uppercase.

```python
s = "python programming"
print(s.upper())
```

## zfill()

**Definition:** Pads the string on the left with zeros until it reaches the specified width.

```python
s = "42"
print(s.zfill(5))
```
