## STRING LOGICAL PROBLEMS PART 2 

#### 1. Take a word and determine whether it is a palindrome. Ignore uppercase/lowercase differences. 

```python
s = input("Enter a string: ")
if s == s[::-1]:
    print("The string is a palindrome")
else:   
    print("The string is not a palindrome")
```

#### 2. Problem: Check whether a sentence is a palindrome while ignoring spaces and case.

```python
s = input("Enter a sentence: ")

s = s.lower().replace(" ", "")

if s == s[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
```

#### 3. Take two words and determine whether they contain the same characters in a different order.

```python
S = input("Enter a string: ")
I = input("Enter a substring: ")

S = S.lower()
I = I.lower()

if sorted(S) == sorted(I):
    print("Anagram")
else:
    print("Not Anagram")
```

#### 4. Take a string and separately count vowels and consonants. Ignore spaces and numbers.

```python
s = input("Enter a string: ")

vowels = 0
consonants = 0

for ch in s.lower():
    if ch.isalpha():
        if ch in "aeiou":
            vowels += 1
        else:
            consonants += 1

print("Vowels:", vowels)
print("Consonants:", consonants)
```

### 5. Count Character Types
#### Count uppercase letters, lowercase letters, digits, spaces, and special characters separately.

```python
s = input("Enter a string: ")

upper = 0
lower = 0
digits = 0
spaces = 0
special = 0

for ch in s:
    if ch.isupper():
        upper += 1
    elif ch.islower():
        lower += 1
    elif ch.isdigit():
        digits += 1
    elif ch.isspace():
        spaces += 1
    else:
        special += 1

print("Uppercase:", upper)
print("Lowercase:", lower)
print("Digits:", digits)
print("Spaces:", spaces)
print("Special:", special)
```

### 6. Extract Only Numbers
#### Problem: Extract all digits from a mixed string

```python
s = input("Enter a string: ")

numbers = ""

for ch in s:
    if ch.isdigit():
        numbers += ch

print(numbers)
```

### 7. Remove Duplicate Characters
#### Problem: Remove duplicate characters while maintaining their original order.

```python
s = input("Enter a string: ")

result = ""

for ch in s:
    if ch not in result:
        result += ch

print(result)
```

### 8. Find Duplicate Characters
#### Problem: Print characters that occur more than once, but print each duplicate only once.

```python
s = input("Enter a string: ")

duplicates = ""

for ch in s:
    if s.count(ch) > 1 and ch not in duplicates:
        duplicates += ch

print("Duplicate characters:", duplicates)
```

### 9. Reverse Each Word
#### Step 1 — Problem: Reverse every individual word without changing the word order.

```python
s = input("Enter a sentence: ")

words = s.split()
result = []

for word in words:
    result.append(word[::-1])

print(" ".join(result))
```
