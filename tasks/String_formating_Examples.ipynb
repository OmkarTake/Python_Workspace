### String formatting examples
1. Employee Salary Table

```python
# Table data using f-string

l1 = [101, 102, 103, 104]
l2 = ["Amit", "Sneha", "Rahul", "Priya"]
l3 = [25000, 32000, 28000, 35000]

r = "ID"
n = "Name"
m = "Salary"
char = 15

print("-" * (char * 3 + 4))
print(f"|{r:^{char}}|{n:^{char}}|{m:^{char}}|")
print("-" * (char * 3 + 4))

for i in range(len(l1)):
    print(f"|{l1[i]:^{char}}|{l2[i]:^{char}}|{l3[i]:^{char}}|")
    print("-" * (char * 3 + 4))
```
