# IF - ELSE + LOGICAL OPERATORS PROBLEMS

10 Python problems covering if, elif, else, comparison operators, and logical operators.

## 01. Positive, Negative, or Zero

**Problem:** Take a number and print Positive, Negative, or Zero.

```python
n = int(input("Enter a number: "))

if n > 0:
    print("Positive")
elif n < 0:
    print("Negative")
else:
    print("Zero")
```

## 02. Even or Odd

**Problem:** Take an integer and print Even or Odd.

```python
n = int(input("Enter a number: "))

if n % 2 == 0:
    print("Even")
else:
    print("Odd")
```

## 03. Voting Eligibility

**Problem:** If age is 18 or above print Eligible to vote, otherwise print Not eligible to vote.

```python
age = int(input("Enter your age: "))

if age >= 18:
    print("Eligible to vote")
else:
    print("Not eligible to vote")
```

## 04. FizzBuzz

**Problem:** Divisible by both 3 and 5 = FizzBuzz; only 3 = Fizz; only 5 = Buzz; otherwise print the number.

```python
n = int(input("Enter a number: "))

if n % 3 == 0 and n % 5 == 0:
    print("FizzBuzz")
elif n % 3 == 0:
    print("Fizz")
elif n % 5 == 0:
    print("Buzz")
else:
    print(n)
```

## 05. Largest of Two Numbers

**Problem:** Take two numbers and print the larger number, or say both are equal.

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

if a > b:
    print(a, "is larger")
elif b > a:
    print(b, "is larger")
else:
    print("Both numbers are equal")
```

## 06. Number Range Check

**Problem:** Check whether a number is between 10 and 50 inclusive using and.

```python
n = int(input("Enter a number: "))

if n >= 10 and n <= 50:
    print("Number is between 10 and 50")
else:
    print("Number is outside the range")
```

## 07. Login Validation

**Problem:** Print Login successful only when username is admin and password is 1234.

```python
username = input("Enter username: ")
password = input("Enter password: ")

if username == "admin" and password == "1234":
    print("Login successful")
else:
    print("Invalid username or password")
```

## 08. Weekend Check

**Problem:** If the day is Saturday or Sunday print Weekend; otherwise print Weekday.

```python
day = input("Enter a day: ")

if day == "Saturday" or day == "Sunday":
    print("Weekend")
else:
    print("Weekday")
```

## 09. Student Result

**Problem:** If both marks are 90 or above print Excellent; if both are 40 or above print Pass; otherwise Fail.

```python
maths = int(input("Enter Maths marks: "))
python = int(input("Enter Python marks: "))

if maths >= 90 and python >= 90:
    print("Excellent")
elif maths >= 40 and python >= 40:
    print("Pass")
else:
    print("Fail")
```

## 10. Simple ATM Validation

**Problem:** Allow withdrawal when amount is greater than 0, within balance, and a multiple of 100.

```python
balance = float(input("Enter balance: "))
withdraw_amount = float(input("Enter withdrawal amount: "))

if withdraw_amount > 0 and withdraw_amount <= balance and withdraw_amount % 100 == 0:
    print("Withdrawal successful")
else:
    print("Invalid withdrawal amount")
```
