## String logical Problems for practice

#### 1. Take a string from the user. Check whether the string is empty or not.

```python
a = input("Enter a string: ")

if a == "":
    print("Empty string")   
else:
    print("Non-empty string")
```

#### 2. Take a string and check whether its length is greater than 5.

```python
a = input("Enter a string: ")
if len(a) > 5:
    print("String length is greater than 5")
else:
    print("String length is not greater than 5")
```

#### 3. Take a string and check whether it is exactly "Python".

```python
a = input("Enter a string: ")
if a == "Python":
    print("String is python")  
else:
    print("String is not python")
```

#### 4. Take a string and check whether it represents "python" regardless of uppercase/lowercase.

```python
a = input("Enter a string: ")
if a.lower() == "python":
    print("String is python")
else:
    print("String is not python")
```

#### 5. Take a string and check whether it starts with "P" and ends with "n". 

```python
a = input("Enter a string: ")
if a.startswith("P") and a.endswith("n"):
    print("String starts with 'P' and ends with 'n'", a)
else:
    print("String does not start with 'P' and end with 'n'")
```

#### 6. Take a sentence and check whether it contains the word "Python".

```python
a  = input("Enter a string: ")
if "Python" in a:
    print("String contains 'Python'")
else :
    print("String does not contain 'Python'")
```

#### 7. Take a string and check whether it does not contain the character "1"

```python
a = input("Enter a string: ")
if "1" not in a:
    print("String does not contain '1'")
else:   
    print("String contains '1'")
```

#### 8. Take a string and check whether it contains only alphabetic characters

```python
a = input("Enter a string: ")
if a.isalpha():
    print("String contains only alphabets")
else:
    print("String contains other characters")
```

#### 9. Take a string and check whether it contains only digits.


```python
a = input("Enter a string: ")
if a.isdigit():
    print("String contains only digits")
else:
    print("String contains other characters")
```

#### 10. A username is valid if its length is at least 5 and it does not contain a space.

```python
a = input("Enter a string: ")
if len(a) >= 5 and a != "":
    print("Valid username")
else:
    print("Invalid username")
```

#### 11. A password is valid if it has at least 8 characters and contains at least one uppercase letter.

```python
a = input("Enter a string: ")
if len(a) >= 8 and a != a.lower():
    print("Valid username")
else:
    print("Invalid username")
```

#### 12. Take a string and determine whether it contains only alphabets, only numbers, or a combination of both. 

```python
a= input("Enter a string: ")
if a.isalnum():
    print("String contains alphanumeric characters")
elif a.isalpha():
    print("String contains only alphabets")
elif a.isdigit():
    print("String contains only digits")
else:
    print("String contains special characters")
```

#### 13. Take two strings and check whether they are equal. Ignore uppercase/lowercase differences.

```python
a = input("Enter a string: ")
b = input("Enter another string: ")
if a.lower() == b.lower():
    print("Both strings are equal (case-insensitive)")
else:
    print("Strings are not equal")
```
